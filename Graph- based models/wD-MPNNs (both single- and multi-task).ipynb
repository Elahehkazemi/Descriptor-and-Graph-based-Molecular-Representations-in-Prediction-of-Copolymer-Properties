```python
from chemprop.train import run_training
from chemprop.args import HyperoptArgs, TrainArgs
from chemprop.hyperparameter_optimization import run_hyperopt

def optimize_hyperparameters(data_path, target_columns, save_dir, is_multitask=False):
    hyperopt_args = HyperoptArgs().parse_args([
        '--data_path', data_path,
        '--dataset_type', 'regression',
        '--target_columns', *target_columns,
        '--num_iters', '20',
        '--config_save_path', f'{save_dir}/best_hyperparams.json',
        '--log_dir', f'{save_dir}/hyperopt_logs',
        '--smiles_column', 'smiles',
        '--metric', 'rmse',
        '--save_dir', save_dir
    ])
    print(f"Running hyperparameter optimization for {'multi-task' if is_multitask else 'single-task'} model...")
    run_hyperopt(hyperopt_args)
    print("Hyperparameter optimization completed.\n")

def train_model(data_path, target_columns, save_dir):
    train_args = TrainArgs().parse_args([
        '--data_path', data_path,
        '--dataset_type', 'regression',
        '--target_columns', *target_columns,
        '--save_dir', save_dir,
        '--config_path', f'{save_dir}/best_hyperparams.json',
        '--gpu', '0'
    ])
    print(f"Training model for targets: {target_columns}")
    run_training(train_args)
    print("Model training completed.\n")

# Single-task
optimize_hyperparameters('single_task_data.csv', ['property'], 'single_task_model')
train_model('single_task_data.csv', ['property'], 'single_task_model')

# Multi-task
optimize_hyperparameters('multi_task_data.csv', ['property1', 'property2', 'property3'], 'multi_task_model', is_multitask=True)
train_model('multi_task_data.csv', ['property1', 'property2', 'property3'], 'multi_task_model')
```
